## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
def dias(semana):
    dias_semana = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]
    if 1 <= semana <= 7:
        return dias_semana[semana - 1]
    else:
        return None

print(dias(2))
````

## Ejercicio 2
En el ejercicio 8 de flujos de control, creábamos una pirámide invertida de números desde el 5. Crea una función que replique el comportamiento de la pirámide, y utiliza un único parámetro de entrada de la función para determinar el número de filas de la pirámide.

````python
def piramide_inver(filas):
    for fila in range(filas, 0, -1):
        for columna in range(fila, 0, -1):
            print(columna, end=" ")
        print()

piramide_inver(10)

````

## Ejercicio 3
Escribe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero sea  mayor que el segundo, o que el segundo sea mayor que el primero

````python
def numeros(numero1: int, numero2: int):
    if numero1 == numero2:
        print("Son iguales")
    elif numero1 > numero2:
        print("El primero es mayor")
    else:
        print("El segundo es mayor")

numeros(2, 2)
````

## Ejercicio 4
Escribe una función que sea un contador de letras. En el primer argumento tienes que introducir un texto, y el segundo que sea una letra que contar. La función tiene que devolver un entero con el número de veces que aparece esa letra, tanto mayúscula, como minúscula

````python
def contador(texto, letra):
    return(texto.count(letra))

texto = "La madre que me pario"
letra = "a"

calculadora = contador(texto, letra)

print(calculadora)
````

## Ejercicio 5
Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string, siendo la clave la letra y el valor el conteo.

````python
def palabra (string):
    dicc = {}

    for i in string:
        if i in dicc:
            dicc[i] += 1 # si la letra ya esta en el dicc se suma 1
        else:
            dicc [i] = 1 # si no está en el diccionario se crea 1
    print(dicc)

string = "estresada"
print(string)
palabra(string)
````

## Ejercicio 6
Escribir una función que añada o elimine elementos en una lista. La función necesita los siguientes argumentos:
* lista: la lista donde se añadirán o eliminarán los elementos
* comando: "add" o "remove"
* elemento: Por defecto es None.

Tendrá como salida la lista

````python
def funcion (lista, comando, elemento = None):
    if comando == "add":
        lista.append(elemento)
    elif comando == "remove":
        lista.remove(elemento)
    else:
        print("Comando erroneo")

    return lista

lista = [52, 63, 98, 78]
comando = "add"
elemento = 34

lista = funcion(lista, comando, elemento)
print(lista)
````

## Ejercicio 7
Crea una función que reciba un número arbitrario de palabras, y devuelva una frase completa, separando las palabras con espacios.

## Ejercicio 8
Escribe un programa que obtenga el enésimo número de la [serie de Fibonacci](https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci). Tienes que crear una función recursiva con un único argumento.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

fibonacci(n) = fibonacci(n-1) + fibonacci(n-2)

````python
def fibonacci(x):
    if x == 1:
        return 1
    elif x == 0:
        return 0
    else:
        return(fibonacci(x-1) + fibonacci(x-2))
````

````python
fibonacci(7)
````
