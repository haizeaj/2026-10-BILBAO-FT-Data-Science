## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

$$radianes = grados*(pi/180)$$

````python
import math
````

````python
grados = float(input("Escribe los grados: "))  
radianes = grados * (math.pi / 180)        
print(f"{grados} grados = {radianes} radianes")    
````

## Ejercicio 6
Mismo programa que en 5, pero en esta ocasión tienen que ser tres inputs y dos salidas. Una de las salidas que nos indique si todos son iguales, y la otra si al menos dos inputs sí que lo son

````python
texto_1 = input("Dime algo: ")
texto_2 = input("Ahora dime lo mismo: ")
otro_mas = input("Y más: ")
if texto_1 == texto_2 and otro_mas == texto_1 :
    print("Son iguales los tres")
else:
    print("No son iguales los tres")
if texto_1 == texto_2 or texto_1 == otro_mas or texto_2 == otro_mas :
    print("Al menos dos son iguales")
else:
    print("Ninguno es igual")
````

## Ejercicio 7
Crea un programa que recoja dos inputs. Tiene que comprobar si su suma es igual, superior o inferior a 10, cada condición por separado

````python
numero1 = float (input ("Dame un numero: "))
numero2 = float(input("Ahora dime otro: "))
sumar = numero1 + numero2
if sumar == 10:
    print("Es igual")
elif sumar > 10:
    print("Es superior")
else:
    print("Es inferior")
````

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "Lo primero de todo, cómo están los máquinas"

1. Pasarlo todo a mayúsculas
2. Pasarlo todo a minúsculas
3. Pasar a mayúsculas solo las iniciales de las palabras
4. Crea una lista dividiendolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`

````python
texto = "Lo primero de todo, cómo están los máquinas"
print(texto.upper())
print(texto.lower())
print(texto.title())
print(texto.split())
print(texto.replace(",", ";"))
print(texto.replace("a", ""))
````
