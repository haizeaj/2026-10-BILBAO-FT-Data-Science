## Ejercicio 4
Realiza un conversor de grados Celsius a Fahrenheit. El conversor tiene que recibir el input del usuario y devolver la conversión.

La conversión de grados sigue la siguiente fórmula: F = C × (9 / 5) + 32 

```python
celsius = float(input("Grados Celsius: "))  # Pide al usuario un número y lo convierte de texto a decimal (float)
fahrenheit = celsius * (9 / 5) + 32         # Aplica la fórmula de conversión de Celsius a Fahrenheit
print(f"{celsius} °C = {fahrenheit} °F")    # Muestra el resultado 
```
