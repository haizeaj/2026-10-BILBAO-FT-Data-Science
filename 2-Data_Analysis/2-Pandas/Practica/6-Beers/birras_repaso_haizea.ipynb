## Ejercicio 24
Teniendo en cuenta la media de amargor de todas las cervezas, obtén una tabla con las cervezas cuyos estados tengan una media de amargor en cerveza por encima de la media

```python

#hecho por la IA

media_global_ibu = birras['ibu'].mean()

# 2. Calcular la media de IBU agrupada por estado
medias_por_estado = birras.groupby('state')['ibu'].mean()

# 3. Filtrar los estados cuya media supere a la media global
estados_altos = medias_por_estado[medias_por_estado > media_global_ibu].index

# 4. Obtener todas las cervezas que pertenecen a esos estados
resultado = birras[birras['state'].isin(estados_altos)]

display(resultado.head())
```
