# Ex1 - Filtering and Sorting Data



### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv) and assign it to a variable called chipo.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")
chipo
```

```python
if chipo['item_price'].dtype == 'object':
    chipo['item_price'] = chipo['item_price'].str.replace('$', '', regex=False).astype(float)
print(chipo['item_price'].head())
print("Tipo de dato actual:", chipo['item_price'].dtype)
```

### Step 3. Name of the max valued product

```python
fila_maxima = chipo['item_price'].idxmax()

articulo_mas_caro = chipo.loc[fila_maxima, 'item_name']
precio_mas_caro = chipo.loc[fila_maxima, 'item_price']


print(f"El producto más caro es: {articulo_mas_caro}")
print(f"Su precio es: ${precio_mas_caro}")
```

### Step 4. How many products cost more than $10.00?

```python
chipo_filtered = chipo.drop_duplicates(['item_name', 'choice_description'])
caros = chipo_filtered[(chipo_filtered['quantity'] == 1) & (chipo_filtered['item_price'] > 10.0)]
print(len(caros))
```

### Step 4.1: Y cuántos pedidos se han hecho con un producto de más de 10$? Es lo mismo?

```python
chipo['product_price'] = chipo['item_price'] / chipo['quantity']
print(len(chipo[chipo['product_price'] > 10.0]))
chipo['product_price'] = chipo['item_price'] / chipo['quantity']
print(chipo[chipo['product_price'] > 10.0]['order_id'].nunique())
```

### Step 4.2: Y cuántos pedidos se han hecho de más de 10$? Es lo mismo?

```python
pedidos_totales = chipo.groupby('order_id')['item_price'].sum()
print(len(pedidos_totales[pedidos_totales > 10.0]))
```

### Step 4.3: Y en cuántos pedidos se ha pagado más de 10$ por un mismo producto? Es lo mismo?

```python
pedidos_con_producto_caro = chipo[chipo['item_price'] > 10.0]['order_id'].nunique()
print(pedidos_con_producto_caro)
```

### Step 5. What is the price of each item and name it unit_price. Get only item_name and unit_price

```python
chipo['unit_price'] = chipo['item_price'] / chipo['quantity']
result = chipo[['item_name', 'unit_price']]
print(result.head())
```

### Step 6. Sort by the name of the item

```python
chipo.sort_values(by='item_name')
```

### Step 7. What was the quantity of the most expensive item ordered? 2 ways

```python
quantity_sort = chipo.sort_values(by='item_price', ascending=False).head(1)['quantity'].values[0]
print(quantity_sort)
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
veggie_salad_orders = chipo[chipo['item_name'] == 'Veggie Salad Bowl']['quantity'].sum()
print(veggie_salad_orders)
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
print(len(chipo[(chipo['item_name'] == 'Canned Soda') & (chipo['quantity'] > 1)]))
```
