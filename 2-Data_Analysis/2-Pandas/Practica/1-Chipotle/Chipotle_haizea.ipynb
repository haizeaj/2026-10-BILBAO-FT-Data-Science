### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv) and assign it to a variable called chipo.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")
chipo
```

```python
chipo.info()
```

```python
chipo["item_price"] = pd.to_numeric(
    chipo["item_price"].astype(str).str.replace("$", ""),
    errors="coerce" # no se para con el error, deja un NaN
)

chipo.info()
```

```python
print(chipo.loc[chipo["item_price"].idxmax(), ["item_name", "item_price", "quantity"]]) #poner siempre el lic, vas por los indices explicitos, por los nombres de las filas
#de la columna price, el mas alto, y del mas alto quiero ver el nombre, el precio y la cantidad
```

```python
#para sber el producto más caro por unidad se divide el item price entre la cantidad.
chipo["unit_price"] = chipo["item_price"] / chipo["quantity"]
chipo.head(10)
```

### Step 3. Name of the max valued product

```python

fila_maxima = chipo['item_price'].idxmax()

articulo_mas_caro = chipo.loc[fila_maxima, 'item_name']
precio_mas_caro = chipo.loc[fila_maxima, 'item_price']


print(f"El producto más caro es: {articulo_mas_caro}")
print(f"Su precio es: ${precio_mas_caro}")
```

### Step 4.2: Y cuántos pedidos se han hecho de más de 10$? Es lo mismo?
