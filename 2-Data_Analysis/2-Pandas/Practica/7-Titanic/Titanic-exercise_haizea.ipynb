31. Using `matplotlib` find the appropriate visualization to show the distribution of the column `'age'`

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 5))

plt.hist(titanic['age'].dropna(), bins=20, edgecolor='black', color='skyblue')

plt.title('Distribución de la Edad (Age)', fontsize=14)
plt.xlabel('Edad', fontsize=12)
plt.ylabel('Frecuencia', fontsize=12)

plt.show()
```

32. Using `matplotlib` find the appropriate plot to visualize the column `'gender'`

```python
gender_counts = titanic['age'].value_counts()
plt.figure(figsize=(6, 4))

plt.bar(gender_counts.index, gender_counts.values, color=['skyblue', 'pink'], edgecolor='black')

plt.title('Distribución por Género', fontsize=14)
plt.xlabel('Género', fontsize=12)
plt.ylabel('Cantidad de pasajeros', fontsize=12)

plt.show()
```

32b. What if you also plot the column `'gender'` using the function [`countplot`](https://seaborn.pydata.org/generated/seaborn.countplot.html) from the library [`seaborn`](https://seaborn.pydata.org/)?

Remember you have never used `seaborn` before, therefore you should install it before importing it.

```python
import seaborn as sns
```

```python
plt.figure(figsize=(6, 4))

# Crea el gráfico de conteo usando countplot de Seaborn
sns.countplot(data=titanic, x='sex', hue='sex', palette=['skyblue', 'pink'], legend=False)

# Añade títulos y etiquetas para mayor claridad
plt.title('Distribución de Pasajeros por Género (Seaborn)', fontsize=14)
plt.xlabel('Género', fontsize=12)
plt.ylabel('Cantidad', fontsize=12)

# Muestra el gráfico
plt.show()
```

33. Using the function [`catplot`](https://seaborn.pydata.org/generated/seaborn.catplot.html) from the library `seaborn`, find out if the hypothesis _"Women are more likely to survive shipwrecks"_ is true or not.

You should get something like this:

![catplot](./img/catplotgen.png)

```python
sns.catplot(data=titanic, x='sex', y='survived', kind='bar', palette='Set2', height=5, aspect=1.2)

# Títulos y etiquetas
plt.title('Tasa de Supervivencia según el Género', fontsize=14)
plt.xlabel('Género', fontsize=12)
plt.ylabel('Tasa de Supervivencia (Media)', fontsize=12)

plt.show()
```

34. Using [`kdeplot`]("https://seaborn.pydata.org/generated/seaborn.kdeplot.html") from `seaborn` represent those who not survived distributed by age.<>

Hint: First you should "filter" the `titanic` dataset where the column "survived" is 0, indexing the column `"age"` only.

Arguments you should pass to the function: 

    - color = "red"
    - label = "Not Survived"
    - shade = True
    
You should get something like this: 

![kdeplot](./img/kdeplotsur.png)

```python
not_survived_age = titanic[titanic['survived'] == 0]['age']

# 2. Creamos el gráfico kdeplot con los argumentos requeridos
sns.kdeplot(not_survived_age, color="red", label="Not Survived", shade=True)

# Añadimos etiquetas y leyenda para completar la visualización
plt.xlabel('Edad', fontsize=12)
plt.ylabel('Densidad', fontsize=12)
plt.title('Distribución de Edad de los No Sobrevivientes', fontsize=14)
plt.legend()

plt.show() # evidentemente los gráficos han sido con ayuda :)
```
