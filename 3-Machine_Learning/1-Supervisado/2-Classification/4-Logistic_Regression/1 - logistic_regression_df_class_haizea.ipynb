## Creamos el modelo

```python
X = np.array(dataframe.drop(columns=['clase'])) #eliminamos la columna que queremos predecir
y = np.array(dataframe['clase'])
```

```python
X = dataframe.drop("clase",axis=1).values# elimina la columna clase, haces lo mismo en las dos celdas
y = dataframe["clase"].values
```

```python
from sklearn.linear_model import LogisticRegression #modelo lineal
```

```python
model = LogisticRegression(max_iter=1000) # entrenamos el modelo de regresion logistica. max_inter= 100 es un parametro de la regresion logistica es para evitar que el modelo diverge, se aleje de la solución, le fijas con esto un máximo de pruebas y pasos para que no se aleje de la respuesta, solución. es una limitación. No sabe si va en la dirección correcta o no, sigue adelante.
#que se pare cuando llega a 1000
model.fit(X, y)
```

```python
predictions = model.predict(X) # despues del fit hacemos el predict, me da las clases
print(predictions)
```

```python
predicions_proba = model.predict_proba(X)
print(np.round(np.array(predicions_proba), 2))
#predicions_proba. aquí me da las probabilidades, cada uno es una predicción, en el caso uno tiene un 0.22% de probabilidad de que sea windows, un 0,09% que sea linus y un 0,69% de que sea mac, por eso se lo asigna
```

```python
model.score(X, y) # el cuadrado del porcentaje de acierto del modelo
```

```python
model.classes_
```

# Validación del Modelo

```python
validation_size = 0.20
seed = 7
X_train, X_test, Y_train, Y_test = model_selection.train_test_split(X,
                                                                    y,
                                                                    test_size=validation_size,
                                                                    random_state=seed)
```

```python
name='Logistic Regression'
kfold = model_selection.KFold(n_splits=10) #Parte los datos en 10 trozos para usar validación cruzada / cross validation. mezclamos 10 veces
cv_results = model_selection.cross_val_score(model, X_train, Y_train, cv=kfold, scoring='accuracy') # 10 veces el accuracy, variando 10 veces los números usados para calcular, al azar
#dentro de cada fold, 10, baraja los datos, guarda datos a train y test saca las metricas y las guarda, así 10 veces 
msg = "%s: %f (%f)" % (name, cv_results.mean(), cv_results.std())
print(cv_results)
print(msg)
# me da una metrica de como de bueno o malo es el modelo, pero no ha mejorado el modelo enlogistic regression da la media de todas las metricas, 0.72 con un margen de error de 0.09
```

```python
model = LogisticRegression(max_iter=1000) # entreno el modelo
model.fit(X_train, Y_train)
```

## Resultados

```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
```

```python
predictions = model.predict(X_test)
print(accuracy_score(Y_test, predictions))
predictions # ha mejorado el accuracy porque hay muy pocos valores con el que calculamos la predicción
```

```python
acierto = accuracy_score(Y_test, predictions) # si valoro con pocos vslores y me guardo uno para test si lo acierto iría la 100% y si lo fallo al 0%

error = 1 - acierto
print("Acierto:", round(acierto*100, 2), "%")
print("Error:", round(error*100, 2), "%")
```

```python
c_matrix = confusion_matrix(Y_test, predictions) # con esta matriz de confución 
```

```python
print(c_matrix)# 16 veces que digo que era windows y lo era, 0 veces que dije que era windows y era linux, y 2 veces que dije que era windows y era MAc
# en la diagonal están los aciertos y los otros lo que predije que no se cumplieron
```

```python
import seaborn as sns
sns.heatmap(c_matrix, annot=True);
```

```python
sns.heatmap(confusion_matrix(Y_test, predictions, normalize='true'), annot=True, 
            fmt='.2%', cmap='Blues'); # cuando dijo Linux predijo el 88,89% de las veces
```

```python
from sklearn.metrics import multilabel_confusion_matrix, precision_score, recall_score, f1_score

multilabel_confusion_matrix(Y_test, predictions)# si era o no windows, windows vs no windows (el primero) linux el 2º y mac el 3º
```
