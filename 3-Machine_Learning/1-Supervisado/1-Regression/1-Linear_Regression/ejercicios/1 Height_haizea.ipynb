### ¿Es una buena técnica para este tipo de problemas? Demuéstralo antes de entrenar un modelo con un análisis descriptivo.

```python
df = pd.DataFrame(lista_alumnos, columns=["nombre", "edad", "altura"])
df
```

### Entrena un modelo de regresión lineal

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
X = df[["edad"]] # con lo que quiero predecir. Podemos tener varias columnas/variables
X
```

```python
y = df["altura"] # lo que quiero predecir. SOlo va a ser una columna
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=4, random_state=42)# que se guarde 4 casos para muestra. COn el random siempre mantiene los datos que aparecen al principio
```

```python
X_test # estos 4 datos no se meten en el fit
```

```python
y_test # estas estaturas que corresponde a las edades de x tampooc entran en el fit
```

```python
lr = LinearRegression() # creop mi modelo de predicción vacío, este modelo es linear regression
```

```python
lr.fit(X_train, y_train) # entrenamos el modelo elegido, se convierte el modelo de este caso
```

```python
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error, mean_squared_error, root_mean_squared_error, r2_score
```

mean_absolute_error es el error medio, el absoluto, cuanto me equivoco de media, aquí son metros
mean_absolute_percentage_error lo mismo pero en porcentaje de error. Da más idea del error y su tamaño
mean_squared_error es el que coge los errores al cuadrado, da el error en unidades al cuadrado, metros al cuadrado. Exagera los errores grandes, para ver algún dato en el que se equivoca mucho.
root_mean_squared_error es la raiz cuadrada. Con el saca la raíz cuadrada que te da con el anterior

```python
y_pred = lr.predict(df_2[['edad']])# queremos guardar en esa variable la columna con la que predigo, no lo que predigo
```

```python
mean_absolute_error(df_2[["altura"]], y_pred) # MAE con el valor real de la estatura con el que yo he predicho, un error medio de:
```

```python
mean_absolute_percentage_error(df_2[["altura"]], y_pred) # una equivocación de un % x medio de todas las estarturas
```

```python
mean_squared_error(df_2[['altura']], y_pred) #MSE exagera los errores grandes , da los errores al cuadrado, si el dato es entre 0 y 1 al elevarlo al cuadrado sale el dato más pequeño
#ver si el modelo tiene errores muy grandes pero media bien o equivocarme un poco más de media y no tener errores muy grandes...
```

```python
root_mean_squared_error(df_2[['altura']], y_pred) ,# le quita el metro cuadrado a la función anterior para mejorar la compransión
```

```python
r2_score(df_2[['edad']], y_pred) # creaste un modelo que predice peor que la media, debería estar entre 0 y 1 para decirte cuan bien predices con tu modelo. 1 perfecto 0 horrible
```
